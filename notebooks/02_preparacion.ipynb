```python
import pandas as pd

df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')

# Numérica que venía como texto
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Clientes sin historia: no han tenido oportunidad de cancelar
df = df[df['tenure'] > 0]

# Etiquetas redundantes con InternetService / PhoneService
cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
        'TechSupport', 'StreamingTV', 'StreamingMovies']
df[cols] = df[cols].replace('No internet service', 'No')
df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')

# Interacción encontrada en la exploración
df['mes_a_mes_fibra'] = ((df['Contract'] == 'Month-to-month') &
                         (df['InternetService'] == 'Fiber optic')).astype(int)

df.to_csv('../data/processed/telco_limpio.csv', index=False)
print(df.shape)
```
